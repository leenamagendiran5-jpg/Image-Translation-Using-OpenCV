# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image using OpenCV
img = cv2.imread("baseball.jpg")
```

```python
rows, cols = img.shape[:2]
M = np.float32([[1, 0, 50],
                [0, 1, 80]])
translated_img = cv2.warpAffine(img, M, (cols, rows))
```

```python
# Display the image using Matplotlib
```

```python
plt.figure(figsize=(10, 8))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(translated_img)
plt.title("Translated Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Image Scaling**

Write a Python program using OpenCV to perform image scaling on the **given image**.

The program should:

1. Scale the image to:
   - **0.5 times** its original size (downscaling)
   - **2 times** its original size (upscaling)
2. Use `cv2.resize` with appropriate scaling factors.
5. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Downscaled Image (0.5×)**  
   - **Upscaled Image (2×)**


```python

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baseball.jpg")
img = img[:, :, ::-1]

small = cv2.resize(img, None, fx=0.5, fy=0.5)
large = cv2.resize(img, None, fx=2, fy=2)

print("Original size:", img.shape)
print("Downscaled size:", small.shape)
print("Upscaled size:", large.shape)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(small)
plt.title("Downscaled (0.5x)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(large)
plt.title("Upscaled (2x)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baseball.jpg")
img = img[:, :, ::-1]

rows, cols = img.shape[:2]

M1 = np.float32([[1, 0.5, 0],
                 [0, 1, 0]])

M2 = np.float32([[1, 0, 0],
                 [0.5, 1, 0]])

horizontal = cv2.warpAffine(img, M1, (cols + int(0.5 * rows), rows))
vertical = cv2.warpAffine(img, M2, (cols, rows + int(0.5 * cols)))

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(horizontal)
plt.title("Horizontally Sheared Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(vertical)
plt.title("Vertically Sheared Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python

import cv2
import matplotlib.pyplot as plt

img = cv2.imread("baseball.jpg")
img = img[:, :, ::-1]

horizontal = cv2.flip(img, 1)
vertical = cv2.flip(img, 0)
both = cv2.flip(img, -1)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(horizontal)
plt.title("Horizontal Reflection")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(vertical)
plt.title("Vertical Reflection")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(both)
plt.title("Both-axis Reflection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("baseball.jpg")

rows, cols = img.shape[:2]

M1 = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rotated_45 = cv2.warpAffine(img, M1, (cols, rows))

M2 = cv2.getRotationMatrix2D((cols/2, rows/2), 90, 1)
rotated_90 = cv2.warpAffine(img, M2, (cols, rows))

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img[:, :, ::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(rotated_45[:, :, ::-1])
plt.title("45 Degree Rotated Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(rotated_90[:, :, ::-1])
plt.title("90 Degree Rotated Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
